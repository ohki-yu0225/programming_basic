## `Series`の使い方

`pandas`では`Series`と`DataFrame`の2種類の配列を使用できる。`Series`は1次元のデータで、`DataFrame`は2次元のデータに対応する。`Series`は`pandas.Series()`関数を用いて、様々な配列から作成することができる。

```python
s1 = pd.Series([0, 1, 2])  # リストから作成
print(s1)

s2 = pd.Series(np.arange(3), index=[3, 4, 5])  # np.arrayから作成、インデックスを引数で指定することができる（デフォルトは自動で0から始まるインデックスが付けられる）
print(s2)

s3 = pd.Series({"千葉県": "ピーナッツ", "大阪府": "たこ焼き", "京都府": "八つ橋"})  # 辞書から作成、辞書の場合はキーがインデックスとなる
print(s3)
```

`Series`はインデックスとデータで構成される。インデックスを`Series`のデータにアクセスするときのキーとして使用することで、他の配列と同様にインデックス式やスライス式を用いたデータの抽出ができる。

```python
# インデックス式
print(s1[0])
print(s3["大阪府"])
```

```python
# スライス式
print(s2[0:2])
print(s3["千葉県":"京都府"])
```

`Series`からインデックスとデータを別々に取り出したいときは`values`プロパティと`index`プロパティを用いる。

```python
print(s3.values)  # インデックス
print(s3.index)  # データ
```

## `DataFrame`の使い方

`DataFrame`は2次元のデータ構造を持つ。`Series`と同様に`pandas.DataFrame()`関数を用いて、各種の配列から作成できる。

```python
d1 = pd.DataFrame([[0,1,2],[3,4,5],[6,7,8]]) #　リストから作成
print(d1)

d2 = pd.DataFrame(np.arange(9).reshape(3, 3), index=["a", "b", "c"], columns=["column_1", "column_2", "column_3"]) # np.arrayから作成、インデックスとカラムを引数で指定することができる（デフォルトは自動で0から始まるインデックス・カラムが付けられる）
print(d2)

d3 = pd.DataFrame({"都道府県": ["千葉県", "大阪府", "京都府"], "主要な観光地": ["ディズニーランド", "大阪城", "清水寺"], "旅行おすすめ度": [3, 4, 5]}) # 辞書から作成、辞書の場合はキーがカラムとなる
print(d3)
```

`DataFrame`から行ごとのデータを参照するときはスライス式が使える。

```python
print(d1[:2]) # d1の最初の2行のデータ
print(d2[-2:]) # d2の終端の2行のデータ
```

列ごとのデータを参照するときは、列名を指定する。

```python
print(d2["column_2"])
print(d2.column_2) # 1列のデータを参照する場合はDataFrame.列名も使える
print(d3[["都道府県", "旅行おすすめ度"]]) # 複数の列のデータを参照する場合は[]に文字列のリストを与える
```

## `iloc`と`loc`

`DataFrame`のデータ参照には`iloc`と`loc`が使える。`iloc`は行と列の位置を指定することで任意の行と列を抽出する。

```python
d = pd.DataFrame(np.random.rand(10, 4), columns=["column_1", "column_2", "column_3", "column_4"])
print(d)
```

```python
print(d.iloc[1])  # dの2行目のデータ
print(d.iloc[2, 2])  # dの3行、3列目のデータ
print(d.iloc[:3])  # dの3行目までのデータ
print(d.iloc[-2:, 1:3])  # dの終端2行、2列目・3列目のデータ
```

`loc`は抽出したい行のインデックスやカラム名を指定することで任意の行と列を抽出する。


```python
print(d.loc[1])  # dのインデックス1のデータ
print(d.loc[2, "column_3"])  # dのインデックス2、カラム"column_3"のデータ
print(d.loc[:3])  # dのインデックス3までのデータ
print(d.loc[8:9, "column_2":"column_3"])  # dのインデックス8・9、カラム"column_2"・"column_3"のデータ
```

インデックスが整数であるとき、`loc`のインデックス参照と`iloc`の位置参照は必ずしも一致しないので注意が必要。

```python
d_drop = d.drop(0)
print(d_drop)
```

```python
print(d_drop.iloc[1])  # 行位置が1（2行目）のデータを抽出
print(d_drop.loc[1])  # インデックスが1（1行目）のデータを抽出
```

## ブールインデックス参照

NumPyでのブールインデックス参照と同様にpandasでも条件式を用いたブールインデックス参照が使える。

```python
idx = d["column_1"] > 0.5  # dのcolumn_1が0.5より大きい行のインデックス
print(idx)
print(d[idx])  # dのcolumn_1が0.5より大きい行のみを抽出
print(d[d["column_1"] > 0.5])  # DataFrame[条件式]の形で1行で書ける
```

```python
print(d[(d["column_1"] > 0.5) & (d["column_2"] > 0.5)])  #「&」を用いて複数の条件式の積集合を抽出できる
print(d[(d["column_3"] > 0.5) | (d["column_4"] > 0.5)])  #「|」を用いて複数の条件式の和集合を抽出できる
```

## データ型

`dtypes`を用いて`DataFrame`の各列のデータ型を調べることができる。pandasで用いられる代表的なデータ型は次の通りである。


|データ型|要素|
| ---- | ---- |
|object|文字列、複数のデータ型の混合|
|int64|整数|
|float64|実数|
|bool|真偽値|

```python
d1 = pd.DataFrame({"object":["A", "B", "C"], "int": [1, 2, 3], "float": [0.1, 0.2, 0.3], "bool": [True, False, True]})
print(d1)
print((d1.dtypes))
```

`astype()`メソッドを用いるとデータの型を変換することができる。

```python
d1_new = d1.astype({"int": "int32"}) # 「int」列のデータ型を32ビット整数に変更
print(d1_new.dtypes)
```

```python
d2 = pd.DataFrame(np.random.rand(400, 5))
print(d2.head(10))  # .head()で冒頭のデータを表示できる
print(d2.dtypes)
```

```python
d2_new = d2.astype("float32")  # 全ての列が実数であるためDataFrameごとに型を変換できる
print(d2_new.head(10))
print(d2_new.dtypes)
```

（参考）32ビットと64ビットでは表現できる桁数や使用するメモリの量に違いがある。64ビットを使用すると扱える桁数が大きくなり、正確な計算が行える一方で、大規模なデータを扱う際は多くのメモリを消費する。pythonではデフォルトでは64ビットが使われる。

```python
print(d2.iloc[0, 0])  # 桁数が大きい
d2.info()  # メモリの使用量が多い
```

```python
print(d2_new.iloc[0, 0])  # 桁数が小さい
d2_new.info()  # メモリの使用量が小さい
```

## データのファイル読み込みと書き出し

`pandas`を用いてデータが記述されているファイルの読み込み・書き出しを行うことができる。csv形式のファイルやpickle形式のファイルがよく使われる。csvファイルの読み込みは`pandas.read_csv()`関数を用いて行う。
